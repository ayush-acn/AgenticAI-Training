## Models

```python
from dotenv import load_dotenv
import os

load_dotenv()

AZURE_OPENAI_API_KEY = os.getenv('AZURE_OPENAI_API_KEY')
AZURE_OPENAI_ENDPOINT = os.getenv('AZURE_OPENAI_ENDPOINT')
```

### Invoking model using - Azure API 

```python
# For uisng the azure openai we use azure openai library

from langchain_openai import AzureChatOpenAI

model = AzureChatOpenAI(
    azure_deployment="gpt-4o",
    api_version = '2024-12-01-preview',
    temperature=0.8
)
```

```python
result = model.invoke('What is the capital of France?')
result.content
```

## Using Azure EntraID

```python
from azure.identity import DefaultAzureCredential, get_bearer_token_provider, InteractiveBrowserCredential

TENANT_ID = os.getenv('TENANT_BACKUP')

token_provider = get_bearer_token_provider(
    DefaultAzureCredential(),
    # credential,
    "https://cognitiveservices.azure.com/.default",
)
```

```python
# Model initialization
llm = AzureChatOpenAI(
    azure_ad_token_provider=token_provider,
    azure_endpoint=AZURE_OPENAI_ENDPOINT,
    azure_deployment="gpt-4o",
    api_version = '2024-05-01',
    temperature=0.8
)
```

```python
try:
    print('sending request')
    response = llm.invoke('What is the capital of France?')
    print(f'Success! Response: {response}')
except Exception as e:
    print(f'Error occurred: {e}')
```
