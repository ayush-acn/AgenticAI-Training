## Messages
Sesding the messages and reciving the responses from the LLMs

*only using the API - OpenAI SDK*

```python
import os
from dotenv import load_dotenv

load_dotenv()

API_KEY = os.getenv("AZURE_OPENAI_API_KEY")
BASE_URL = os.getenv("AZURE_OPENAI_BASE_URL")
```

```python
from langchain_openai import ChatOpenAI


model = ChatOpenAI(
    base_url = BASE_URL,
    api_key = API_KEY,
    model = "gpt-4o",
    temperature = 0.8
)
```

### Single Messages

```python
messages = [
    (
        "system",
        "You are a helpful language translation assistant that translates English to French. Translate the user sentence.",
    ),
    ("human", "I love python programming."),
]

ai_msg = model.invoke(messages)
ai_msg.content
```

```python
# Directly sending and prnitng the result 
print(model.invoke('Translate this to hindi - I love python programming.').content)
```

```python
# Using Classes

from langchain_core.messages import HumanMessage, SystemMessage, AIMessage

messages = [
    SystemMessage(content="Solve the following math problems"),
    HumanMessage('what is the square root of 81 divided by 9')
]

ai_msg = model.invoke(messages)
print(ai_msg.content)
```

### Multiturn Conversation

```python
messages = [
    SystemMessage(content="Solve the following math problems"),
    HumanMessage('what is the square root of 81 divided by 9'),
    AIMessage('Compute step by step:\n\n1. Square root of 81 is 9.  \n2. Then 9 divided by 9 is 1.\n\nAnswer: 1'),
    HumanMessage('What if I say the correct answerr is 3.')
]

ai_msg = model.invoke(messages)
print(ai_msg.content)
```

```python
# Directly callingthe model

response = model.invoke([
    (
        "system",
        "You are a helpful assistant.",
    ),
    (
        "user",
        "I am going to Paris, what should I see?",
    ),
    (
        "assistant",
        "Paris, the capital of France, is known for its stunning architecture, art museums, historical landmarks, and romantic atmosphere. Here are some of the top attractions to see in Paris:\n\n1. The Eiffel Tower: The iconic Eiffel Tower is one of the most recognizable landmarks in the world and offers breathtaking views of the city.\n2. The Louvre Museum: The Louvre is one of the worlds largest and most famous museums, housing an impressive collection of art and artifacts, including the Mona Lisa.\n3. Notre-Dame Cathedral: This beautiful cathedral is one of the most famous landmarks in Paris and is known for its Gothic architecture and stunning stained glass windows.\n\nThese are just a few of the many attractions that Paris has to offer. With so much to see and do, its no wonder that Paris is one of the most popular tourist destinations in the world.",
    ),
    (
        "user",
        "What is so great about #1?",
    )
])

print(response.content)
```

## From Foundary

```python
from langchain_openai import AzureChatOpenAI
subscription_key=os.getenv("AZURE_INFERENCE_KEY")

model = AzureChatOpenAI(
    api_key=subscription_key,
    azure_endpoint="https://uttamfoundry01.cognitiveservices.azure.com/",
    azure_deployment="gpt-4o",  # or your deployment
    api_version="2024-12-01-preview",  # or your api version
    temperature=2
)

result = model.invoke("What is the capital of India")
print(result.content)
```
