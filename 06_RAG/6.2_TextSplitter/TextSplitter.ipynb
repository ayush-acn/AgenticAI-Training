## Text Splitting

Splitting the text and making the chunks, for storing it in the DB.

https://docs.langchain.com/oss/python/integrations/splitters

### Setup

```python
import os
from dotenv import load_dotenv
load_dotenv()

API_KEY = os.getenv('AZURE_OPENAI_API_KEY')
BASE_URL = os.getenv('OPENAI_BASE_URL')

os.environ['LANGCHAIN_API_KEY'] = os.getenv('LANGSMITH_API_KEY')
os.environ["LANGCHAIN_TRACING_V2"] = "true"
os.environ['LANGSMITH_PROJECT'] = 'AgenticAITraining'
```

```python
from langchain_openai import ChatOpenAI
from langchain_core.prompts import ChatPromptTemplate
from langchain_core.output_parsers import StrOutputParser

prompt = ChatPromptTemplate.from_template(
    template=''
)

model = ChatOpenAI(
    api_key = API_KEY,
    base_url=BASE_URL,
    model='gpt-5.1',
    temperature=1.5
)

parser = StrOutputParser()

chain = (prompt | model | parser).with_config({'run_name':'SplittingChain'})
```

### Char Count

```python
from langchain_text_splitters import RecursiveCharacterTextSplitter

text = text = """
The quick brown fox jumps over the lazy dog.
This is a single paragraph.

Another paragraph starts here.
This one has more content and is also a single paragraph. It is longer than the first one to demonstrate the splitting.
You can see that there are two paragraphs here.
They are separated by a double newline.

A third paragraph follows.

A fourth paragraph is here to make the text even longer and to show how the splitter handles more content. It should be split into multiple chunks based on the chunk size specified.
"""
```

```python
text_splitter = RecursiveCharacterTextSplitter(
    chunk_size = 100,
    # chunk_overlap = 13
    chunk_overlap = 0
)

chunks = text_splitter.split_text(text)
print(len(chunks))
```

```python
for i, chunk in enumerate(chunks):
    print(f'Chunk {i}:\t {chunk}')
```

```python
text_splitter = RecursiveCharacterTextSplitter(
    chunk_size = 100,
    chunk_overlap = 12
)

chunks = text_splitter.split_text(text)
for i, chunk in enumerate(chunks):
    print(f'Chunk {i} - {len(chunk)}:\t {chunk}')
```

### Token Count

```python
text_splitter = RecursiveCharacterTextSplitter.from_tiktoken_encoder(
    encoding_name='o200k_base',
    chunk_size = 100,
    chunk_overlap = 0
)

chunks = text_splitter.split_text(text)
print(len(chunks))
```

```python
for i, chunk in enumerate(chunks):
    print(f'Chunk {i} - {len(chunk)} - :\t {chunk}')
```
