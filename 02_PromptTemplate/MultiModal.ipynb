## MultiModal

Using MultiModal - Sending Images and Texts togather

```python
import os
from dotenv import load_dotenv
load_dotenv()

API_KEY = os.getenv('AZURE_OPENAI_API_KEY')
BASE_URL = os.getenv('OPENAI_BASE_URL')
```

```python
from langchain_openai import ChatOpenAI

model = ChatOpenAI(
    base_url = BASE_URL,
    api_key = API_KEY,
    model = 'gpt-5.1',
    temperature = 0.9    
)
```

### Sending Image via URL

```python
from langchain_core.messages import HumanMessage

image_url = 'https://wallpapercave.com/wp/wp3049325.jpg'

multimodal_content = [
    {
        'type': 'image_url',
        'image_url': {'url': image_url}
    },
    {
        'type': 'text',
        'text': 'What is this image about? Give me a description of the content.'
    }
]


message = HumanMessage(content = multimodal_content)
message
```

```python
try:
    response = model.invoke([message])
    print(response.content)
except Exception as e:
    print(f"An error occurred: {e}")
```

```python
try:
    model2 = ChatOpenAI(base_url=BASE_URL, api_key=API_KEY, model='gpt-4o', temperature=0.9)
    response = model2.invoke([message])
    print(response.content)
except Exception as e:
    print(f"An error occurred: {e}")
```

### Sending Image form the local Storage

```python
# Libraries to process the images
import base64
import mimetypes
```

```python
def encode_image(image_path):
    with open(image_path, 'rb') as f:
        image_data = f.read()
    encoded_image = base64.b64encode(image_data).decode('utf-8')
    mime_type, _ = mimetypes.guess_type(image_path)
    return {
        'data': encoded_image,
        'mime_type': mime_type
    }
```

```python
file_path = './images/car.jpg'

enoder_resp = encode_image(file_path)
encoded_img = enoder_resp['data']
mime_type = enoder_resp['mime_type']

supported_types = ['image/jpeg', 'image/png', 'image/gif', ' image/webp']

if mime_type not in supported_types:
    exit()
```

```python
multimodal_content = [
    {
        'type': 'image',
        'base64': encoded_img,
        'mime_type': mime_type
    },
    {
        'type': 'text',
        'text': 'What is this image about? Give me a description of the content.'
    }
]

message = HumanMessage(content = multimodal_content)
message
```

```python
try:
    response = model.invoke([message])
    print(response.content)
except Exception as e:
    print(f"An error occurred: {e}")
```
