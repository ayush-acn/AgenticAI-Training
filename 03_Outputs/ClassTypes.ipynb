### TypedDict
Static typing not validation

This is a dictonary declared as a calls

```python
from typing import TypedDict

class Person(TypedDict):
    name: str
    age: int
    city: str
```

```python
p1 = Person(name= 'Ayush', age= 24, city= 'Gurgaon')

print(p1)
print(type(p1))
```

```python
p2 = Person(name = 'John', age = 30)

print(p2)
print(type(p2))
p2.keys()
```

### Data Class

Pythonic standard for plain data object - Need to define all the elements


- This is similar to TypedDict but instead of dict this is a class

```python
from dataclasses import dataclass

@dataclass
class Person:
    name: str
    age: int
    city: str = None
```

```python
p1 = Person(name= 'Ayush', age= 24, city= 'Gurgaon')
print(p1)
print(type(p1))
p1.name
```

```python
p1.city = 'Delhi'
print(p1)
print(p1.city)
```

```python
p2 = Person(name= 'John', age= 'Thirtry')
print(p2)
print(type(p2))
print(list(vars(p2)))
```

### Typing Annotation

```python
from typing_extensions import Dict
from typing import List, Literal, Annotated, Optional
```

```python
class DataClassReview:
    summary: str
    sentiment: Literal['positive', 'negative', 'neutral']
```

```python
r1 = DataClassReview()
r1.summary = "This product is amazing!"
r1.sentiment = 'kuchg bi'
r1.user = 'test'
print(r1)
print(type(r1))
print(list(vars(r1)))
r1.sentiment, r1.user, type(r1.sentiment), r1
```

### Pydantic

For validation

- Default values, optional values, coerece ...

```python
from pydantic import BaseModel, Field, ValidationError
from typing import Optional, List
```

```python
class studnet(BaseModel):
    name: str
    age: int
    city: str
```

```python
s1 = studnet(name= 'Ayush', age= 24, city= 'Gurgaon')
print(s1)
print(type(s1))
```

```python
class studnet(BaseModel):
    name: str
    age: Optional[int] = Field(default=None, description="Age of the student")  
    city: Optional[List[str]] = Field(default_factory=list, description="City of the student")
```

```python
s1 = studnet(name= 'Ayush', age= 24, city= ['Gurgaon'])
print(s1)
print(type(s1))
print(dict(s1))

s2 = studnet(name= 'Aman')
print(s2)
print(type(s2))
```
